Extracting some info

```python
import pandas as pd
df= pd.read_csv("homework5.csv")
df.head () # this is to show the first rows
```

Which exam is hardest and which is easiest?

```python
import matplotlib.pyplot as plt
avg_hours = df.groupby("Exam") ["Score"].mean()

#Column chart: Average score by Exam
avg_hours.plot(kind="bar" , x="Exam" , y= "Score")
plt.title("Average Score by Exam")
plt.xlabel("Exam")
plt.ylabel("Average Score")

plt.show()
```

The chart shows that students scored highest in English, while Math and History averages were slightly lower. This suggests English may be easier for the group, while Math and History are more challenging.

```python
# Average score per student
avg_scores = df.groupby("Student")["Score"].mean().sort_values(ascending=False)

plt.figure(figsize=(8, 8))

# Pie chart: Average score share per student
avg_scores.plot(kind="pie", legend=False, autopct='%1.1f%%')

plt.title("Who Performs Best (Average Score per Student)")
plt.ylabel("")  # remove the default 'Score' label
plt.show()
```

The pie chart shows how each student contributes to overall performance. Some have have larger slices, meaning they achieved higher average scores, while others have smaller slices indicating they didn't do that well.

```python
plt.scatter(df["Age"], df["Score"], s=df["Age"]*30, alpha=0.6)
plt.title ("Age vs Score")
plt.xlabel("Age")
plt.ylabel("Score")
plt.show
```

The scatter plot shows that students aged 19–22 achieve a wide range of scores. There is no clear link between age and performance, since both high and low results appear at every age. This suggests that age does not strongly influence exam success in this group.Also, the bubbles are made partly transparent, which makes overlapping points easier to see and the chart easier to understand.

```python
# Calculate quartiles and extremes for Score
score_min = df["Score"].min()
score_q1 = df["Score"].quantile(0.25)
score_median = df["Score"].median()
score_q3 = df["Score"].quantile(0.75)
score_max = df["Score"].max()

print("Minimum:", score_min, 
      "Q1:", score_q1, 
      "Median:", score_median, 
      "Q3:", score_q3, 
      "Maximum:", score_max)

plt.figure(figsize=(8,6))
df[["Score"]].boxplot()
plt.title("Distribution of Student Scores")
plt.show()
```

The box plot shows that English scores are generally higher and more consistent, while Math and History have lower medians and more variation. This suggests students perform more evenly in English than in the other subjects.
