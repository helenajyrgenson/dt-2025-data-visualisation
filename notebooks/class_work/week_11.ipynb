```python
sentence = "mulonniipaljuteha" #when counting first letter isn't one but zero
#extract first three letter
print("Print first three letters:", sentence [0:3]) #start: end
print (sentence [:3])

#extract letter from five to nine
print("Letters from five to nine:", sentence [5:9])
#Extract letters from 5th to end
print(sentence [5:])
#Extract the last character
print(sentence [-1])
#Extract last 4 characters
print(sentence [-4:])
#Extract first letter of the substring starting from 5
print(sentence [5:][0])
#Extract first five letters of the substring starting from 5
print (sentence [5:][0:5])
#Extract entire string/text
print(sentence [:])
#Extract every 2nd character
print(sentence[::2])
```

```python
sentence = "Alice went to the park and saw a beautiful bird singing"

#Split this sentence into different words, ""this in there and just a space between
words = sentence.split(" ")

print(words)

#Extract the first word
print(sentence[0:5])

#Extract the first three words
print(words[0:3])

#Extract last three words
print(words[-3:])

#Find out how many words are there
print("The number of words:", len(words))
```

```python
#get text from file, make it into wordcloud and change the shape
from wordcloud import WordCloud
from PIL import Image #import image class from PIL
import urllib.request #Library to work with URL

import numpy as np #another library, in order to work with image masks we need this library

#text link from the drive https://drive.google.com/file/d/11aaxlQdcjTygtzvzCBAIuQ16t8eLwyXi/view?usp=sharing , use same 
#format as before https://drive.google.com/uc?export=download&id=FILE_ID 
text = urllib.request.urlopen("https://drive.google.com/uc?export=download&id=11aaxlQdcjTygtzvzCBAIuQ16t8eLwyXi").read().decode()

#also gonna need the image, just copy previous one
img = urllib.request.urlretrieve("https://drive.google.com/uc?export=download&id=1sLctU_YlWYfq5jYNOm5yMOg7k8ADDsSc", "image1.png")

#gonna recreate the wordcloud and this mask=...and so on is to change the shape and also gonna change the background color
WordCloud(mask=np.array(Image.open("image1.png")), background_color="beige").generate_from_text(text).to_image()
```
