```python
import pandas as pd 

df= pd.read_csv("sports.csv")
df.head () # this is to show the first rows
```

```python
df.describe() # generate some statistics
```

Box plot for each athlete to find the number of years of practise 

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice")
plt.show
```

Without legend

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice", legend=False)
plt.show
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice")
plt.ylabel("Years")
plt.show
```

Last one was to get the "Years" there

To get years of practise per athlete

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Years_of_practice")
plt.ylabel ("Years")
plt.title("Years of practise per athlete")
plt.show
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar", x="Name", y="Age")
plt.ylabel ("Age of each athlete")
plt.title("Age")
plt.show
```

Compare age and years of practise

```python
plt.scatter(df["Age"], df["Years_of_practice"])
plt.show
```

Bar chart. Average age per sport

```python
avg_age=df.groupby("Sport") ["Age"].mean()
avg_age.plot(kind="bar")
plt.show()
```
