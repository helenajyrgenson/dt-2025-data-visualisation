```python
import urllib.request

# URL of text file (poem used in class)
url = "https://raw.githubusercontent.com/helenajyrgenson/dt-2025-data-visualisation/refs/heads/codespace-jubilant-orbit-gr97595599x39g76/notebooks/class_work/poem1.txt"

text = urllib.request.urlopen(url).read().decode("utf-8")

# Print first 200 characters
print("FIRST 200 CHARACTERS:")
print(text[:200])
```

```python
# Split text into words
words = text.split()

print("BASIC INFO")
print("Number of characters in text:", len(text))
print("Number of words in text:", len(words))

print("FIRST 10 WORDS:")
print(words[:10])

print("LAST 10 WORDS:")
print(words[-10:])
```

```python
# Slicing text

print("STRING SLICING")
print("First 50 characters:")
print(text[:50])

print("Last 50 characters:")
print(text[-50:])

print("Every 2nd character from first 50:")
print(text[:50:2])
```

```python
# Looping over words and length

print("WORD LENGTHS (FIRST 10)")
for word in words[:10]:
    print(word, "- length:", len(word))
```

In task 1, I read a text file from a URL (poem we used in class, I used the same one, hope that's okay) and printed the first 200 words of it. Then I split the text into words, counted how many words there were and sliced the text, so it gave me the first and last characters. I also printed the length of the first ten words using a loop.

```python
from wordcloud import WordCloud

# Basic word cloud
wc = WordCloud().generate(text)
img = wc.to_image()
img.show()
```

```python
wc2 = WordCloud(stopwords=["and", "then", "but", "to", "on"], background_color="white").generate(text)
img2 = wc2.to_image()
img2.show()


```

In task 2, I generated two wordclouds from the same text. In the first cloud (black background) all words were included, which made short and common words very large. These do not tell much about the meaning of the text, only which words appear the most often.

In the second cloud, I used English stopwords (and, then, but, to, on) to remove some of the most repetitive connector-words. Because I only filtered out those specific words, others like "the" still remained and appeared very large. After filtering these out, more meaningful words became clearer. The biggest words in the cloud were Vanemuine, people, play, kannel, forest, meadow and Estonian.
Cloud  helps understand what the poem is mainly about: Estonian folklore, people, nature, meadows, forests, music (especially the kannel), and storytelling.
The cloud gives a quick visual summary of the text without needing to read the text.
